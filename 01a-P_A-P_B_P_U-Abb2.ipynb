## 🎲 Gewinnwahrscheinlichkeiten beim Ein-Würfel-Fall

Dieses Programm berechnet **exakt** (mittels Bruchrechnung) und **numerisch gerundet** die Gewinnwahrscheinlichkeiten  
$$ P_A(V,W),\ P_B(V,W),\ P_U(V,W) $$  
für zwei vorgegebene **Setzstrategien** $V$ und $W$.

---

### ✅ Voraussetzungen

- Es wird genau **ein Würfel** mit benutzerdefinierten Wahrscheinlichkeiten verwendet.
- Die Strategien $V$ und $W$ geben an, wie viele Chips auf die jeweiligen Felder gesetzt wurden.

---

### ⚙️ Eingaben (am Ende des Programms)

- `p_list`: die Wahrscheinlichkeiten für die einzelnen Felder.  
- `V, W`: die beiden Setzstrategien


```Python
# Beispiel 1   (3 Felder, 6 Chips):
p_list = (Fraction(1,2), Fraction(1,3), Fraction(1,6))
V = (3,2,1)
W = (4,2,0)
```

```Python
# Beispiel 2   (6 Felder, 18 Chips):
p_list = (Fraction(3,18), Fraction(5,18), Fraction(4,18),
          Fraction(3,18), Fraction(2,18), Fraction(1,18))
V = (3,5,4,3,2,1)
W = (3,7,4,3,1,0)
```
> 📌 Die Beispiele können in das Programm auch kopiert werden.
---

### ▶️ So geht’s weiter

In der **nächsten Code-Zelle** befindet sich das zugehörige Python-Programm.  
Dort müssen lediglich folgende Eingaben angepasst werden:

- `p_list`: Wahrscheinlichkeiten für die Felder (z. B. `p_list = (Fraction(1,2),Fraction(1,3),Fraction(1,6))`
- `V`, `W`: Setzstrategien der beiden Spieler (z. B. `V = (3, 2, 1)`)

> 📌 Achten Sie darauf, dass `p`, `V` und `W` gleich lang sind!

Nach wenigen Sekunden werden die folgenden exakten und gerundeten Ergebnisse als Tupel $(P_A(V,W), P_B(V,W),P_U(V,W))$ ausgegeben:

- $P_A(V,W)$ – A gewinnt
- $P_B(V,W)$ – B gewinnt
- $P_U(V,W)$ – Unentschieden

📌 **Hinweis:** `Fraction(a, b)` bedeutet $\displaystyle\frac{a}{b}$;  Beispiel: `Fraction(1181921, 2400000)` = $\displaystyle\frac{1181921}{2400000}$


📎 Viel Erfolg beim Experimentieren mit eigenen Strategien!

---

````python
from fractions import Fraction
from functools import lru_cache

@lru_cache(None)
def P_A(V, W):
    V, W = list(V), list(W)
    sum_V, sum_W = sum(V), sum(W)
    if sum_V == 0 and sum_W == 0: return 0 # unentsch.
    elif sum_V == 0: return 1   # A gewinnt
    elif sum_W == 0: return 0   # B gewinnt

    s = sum(p for p, v_i, w_i in zip(p_list, V, W) if v_i or w_i)

    prob = 0
    for i, p in enumerate(p_list):
        if V[i] or W[i]:
            V_new, W_new = V.copy(), W.copy()
            if V_new[i]: V_new[i] -= 1
            if W_new[i]: W_new[i] -= 1
            prob += (p/s)*P_A(tuple(V_new),tuple(W_new))
    return prob

def compute_outcomes(V, W):
    PA = P_A(tuple(V), tuple(W))
    PB = P_A(tuple(W), tuple(V))
    PU = 1 - PA - PB
    return PA, PB, PU

# --- Eingabe von p, S und T sowie eine Ausgabe ---
p_list = (Fraction(1,2), Fraction(1,3), Fraction(1,6))
V = (3,2,1)
W = (4,2,0)

PA, PB, PU = compute_outcomes(V, W)
print(f"P_A({V},{W}) = {PA} ≈ {float(PA):.4f}")
print(f"P_B({V},{W}) = {PB} ≈ {float(PB):.4f}")
print(f"P_U({V},{W}) = {PU} ≈ {float(PU):.4f}")
````
