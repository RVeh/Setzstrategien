```python
import PropSetzstrategien as ps
import itertools
from functools import lru_cache
 
# ============== Eingabe ====================   

p_list = [6/36, 10/36, 8/36, 6/36, 4/36, 2/36]
base = (3,5,4,3,2,1)  # Vergleichsstrategie V gegen alle anderen
total_chips = sum(base)
min_limits = (1, 3, 2, 1, 1, 0)  # Einschränkungen MIN für die Fächer
max_limits = (7, 9, 8, 7, 4, 2)  # Einschränkungen MAX für die Fächer

# ============== Berechnungen ===============   
# Vollständige Suche (mit MIN/MAX)
better = ps.find_better_strategies(base, total_chips, min_limits, max_limits, p_list)

# ============== Ausgabe ==================== 
print(fr"Einschrängungen: MIN: {min_limits};   MAX: {max_limits}")
print()
if better:
    print(fr"Bessere Strategien als V={base}:")
    for strat, diff in better[:50]:   # die ersten 50, falls vorhanden
        print(f"Strategie {strat} gewinnt um {diff:.4f}")
else:
    print("Keine bessere Strategie gefunden.")
```
