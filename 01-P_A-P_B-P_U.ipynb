## 🎲 Gewinnwahrscheinlichkeiten beim Ein-Würfel-Fall

Dieses Programm berechnet **exakt** (mittels Bruchrechnung) und **numerisch gerundet** die Gewinnwahrscheinlichkeiten  
$$ P_A(V,W),\ P_B(V,W),\ P_U(V,W) $$  
für zwei vorgegebene **Setzstrategien** $V$ und $W$.

---

### ✅ Voraussetzungen

- Es wird genau **ein Würfel** mit benutzerdefinierten Wahrscheinlichkeiten verwendet.
- Die Wahrscheinlichkeiten sollten sich zu $1$ summieren.
- Die Strategien $V$ und $W$ geben an, wie viele Chips auf die jeweiligen Felder gesetzt wurden.

---

### ⚙️ Eingaben

- `p`: die Wahrscheinlichkeiten für die einzelnen Felder.  
- `V, W`: die beiden Setzstrategien
   
---
### 🧩 Modul `PropSetzstrategien.py`

Die Datei `PropSetzstrategien.py` enthält **alle zentralen Funktionen** zur Berechnung:

- Gewinnwahrscheinlichkeiten (`P_A`, `P_B`, `P_U`)
- Strategievergleiche
- Simulationen und Konfidenzintervalle
- Visualisierungen

> 📌 **Wichtig:** Diese Datei muss sich im **Hauptverzeichnis** befinden,  
> damit sie in den Notebooks importiert werden kann (z. B. `import PropSetzstrategien as ps`).

Man kann `PropSetzstrategien.py` als **Funktionssammlung** bezeichnen – sie wird von allen Jupyter-Notebooks gemeinsam genutzt.

### ▶️ So geht’s weiter

In der **nächsten Code-Zelle** befindet sich das zugehörige Python-Programm.

- `p`: Wahrscheinlichkeiten für die Felder (z. B. `p = (1/2, 1/3, 1/6)`)
- `V`, `W`: Setzstrategien der beiden Spieler (z. B. `V = (3, 2, 1)`)

> 📌 Achte Sie darauf, dass `p`, `V` und `W` gleich lang sind!

Nach wenigen Sekunden werden die folgenden exakten und gerundeten Ergebnisse als Tupel $(P_A(V,W), P_B(V,W),P_U(V,W))$ ausgegeben:

- $P_A(V,W)$ – A gewinnt
- $P_B(V,W)$ – B gewinnt
- $P_U(V,W)$ – Unentschieden

```python
from fractions import Fraction
import PropSetzstrategien as ps

"""
PropSetzstrategien:
- Legt p, V, W fest
- berechnet P_A, P_B, P_U (exakt mit Fraction, oder numerisch)
- gibt Ergebnisse formatiert aus
"""

# ========= Eingaben =================

# Eingabe der Wahrscheinlichkeiten
p = (1/2,1/3,1/6)

# Eigabe der Setzstrategien V und W
V = (3,2,1)
W = (4,2,0)

# ========Sicherheits-Check:===========

#  `exact=False`: falls numerisch
ps.p_list = ps.prepare_p_list(p, exact=True) 
assert len(ps.p_list) == len(V) == len(W), \
"p_list, V und W müssen alle dieselbe Länge haben!"

# =========== Berechnung =============

PA, PB, PU = ps.compute_outcomes(V, W)

# =========== Ausgabe ================
"""
Formatiert und druckt Strategien, p_list
und die exakten sowie numerischen Ergebnisse.
Bsp.: Fraction(2,3)=2/3
"""
print("")
print("-" * 75)
print(f"Strategien: V = {V}, W = {W}")
print(f"Wahrscheinlichkeiten p_i: {ps.p_list}")
print("-" * 75)
print("Ergebnisse (P_A(V,W), P_B(V,W), P_U(V,W)):\n")

# exakte Brüche
print("exakt:", (PA, PB, PU),"\n")
# numerisch gerundet
print("numerisch:", 
      f"({float(PA):.4f}, {float(PB):.4f}, {float(PU):.4f})")
print("-" * 75)
```
