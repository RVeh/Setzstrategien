# 🧠 Einstieg: Setzstrategien und Gewinnwahrscheinlickeiten

Willkommen im interaktiven Projekt zur Analyse von Chip-Verteilstrategien!  
Hier können Sie mithilfe von Python und mathematisch fundierten Funktionen:

- Wahrscheinlichhkeiten für zwei Setzstrategien berechnen (ein-und zwei-Würfel-Fall) 
- alle möglichen Strategien systematisch vergleichen
- grafisch auswerten 
- Grundlagen & Herleitungen nachlesen

---

Wählen Sie ein Thema aus, um direkt loszulegen:
