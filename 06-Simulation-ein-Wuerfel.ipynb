```python
from fractions import Fraction
import PropSetzstrategien  as ps

# ===========. Eingabe =========================

p=(1/2, 1/3, 1/6)  # Wahrscheinlichkeiten 

V = (3, 2, 1)      # Setzstrategie 1
W = (4, 2, 0)      # Setzstrategie 2

sim=10_000   # Anzahl an Simulationen

# ============ Sicherheitsabfrage ==============

# Sicherheits-Check
#`exact=False`: falls numerisch
ps.p_list = ps.prepare_p_list(p, exact=True) 
assert len(ps.p_list) == len(V) == len(W), \
"p_list, V und W müssen alle dieselbe Länge haben!"

# === Berechnung der Simulation mit WALD-KI ====

# Mit Wald-CI
cis = ps.simulate_with_wald(p, V, W, N=sim, alpha=0.05)
pA_hat, pB_hat, pU_hat = cis['P_A'][0], cis['P_B'][0], cis['P_U'][0]
pA_low, pB_low, pU_low = cis['P_A'][1][0], cis['P_B'][1][0], cis['P_U'][1][0]
pA_high, pB_high, pU_high = cis['P_A'][1][1], cis['P_B'][1][1], cis['P_U'][1][1]

# ============== Ausgabe =======================

print(f"\nV={V}; W={W}")
print(f"p_i={ps.p_list}")
print(f"{sim} Simulationen")
print("-"*60)
print(f"P_A_Schätzung = {pA_hat:.4f}, 95%-CI = [{pA_low:.4f}, {pA_high:.4f}]")
print(f"P_B_Schätzung = {pB_hat:.4f}, 95%-CI = [{pB_low:.4f}, {pB_high:.4f}]")
print(f"P_U_Schätzung = {pU_hat:.4f}, 95%-CI = [{pU_low:.4f}, {pU_high:.4f}]")
```
