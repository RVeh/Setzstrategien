## 🎲 Schätzwerte für Gewinnwahrscheinlichkeiten

Dieses Programm ermittelt mithilfe von **Simulationen** Schätzwerte für die zugehörigen Gewinnwahrscheinlichkeiten $$ P_A(V,W),\ P_B(V,W),\ P_U(V,W) $$ und gibt auch jeweils die zugehörigen 95%-WALD-Konfidenzintervalle aus. Es werden der **ein- und zwei-Würfel-Fall** betrachtet. Es kann hiermit untersucht werden, welche Auswirkungen es hat, mit einem oder mit zwei Würfel zu spielen. Für den **ein-Würfel-Fall** werden zusätzlich die exakten Wahrscheinlichkeiten berechnet und ausgegeben. Somit kann auch untersucht werden, ob die eigenlich unbekannten Wahrscheinlichkeiten von den zugehörigen Konfidenzintervallen überdeckt werden.

📌 **Bemerkung:** Simulationsergebnisse sollten eigenlich immer mit Konfidenzintervallen angegeben werden.

---

### ✅ Voraussetzungen

- Es werden die beiden Fälle **ein Würfel** und **zwei Würfel** mit benutzerdefinierten Wahrscheinlichkeiten betrachtet.
- Die Wahrscheinlichkeiten sollten sich zu $1$ summieren.
- Die Strategien $V$ und $W$ geben an, wie viele Chips auf die jeweiligen Felder gesetzt wurden.

---

### ⚙️ Eingaben

- `p`: die Wahrscheinlichkeiten für die einzelnen Felder.  
- `V, W`: die beiden Setzstrategien
-  `sim`: Anzahl an Simulationen

- `p`: Wahrscheinlichkeiten für die Felder (z. B. `p=(3/18, 5/18, 4/18, 3/18, 2/18, 1/18)`
- `V`, `W`: Setzstrategien der beiden Spieler (z. B. `V = (3, 5, 4, 3, 2, 1), W = (3, 7, 4, 3, 1, 0)`)
-  `sim`: Anzahl an SImulationen (z.B. `sim=10000`)

```python
from fractions import Fraction
import PropSetzstrategien as ps

# =========== Eingabe ==================

p = (3/18, 5/18, 4/18, 3/18, 2/18, 1/18)

V = (3, 5, 4, 3, 2, 1)
W = (3, 7, 4, 3, 1, 0)

sim=10_000

# =======Sicherheits-Check =============
#`exact=False`: falls numerisch

ps.p_list = ps.prepare_p_list(p, exact=True) 
assert len(ps.p_list) == len(V) == len(W), \
"p_list, V und W müssen alle dieselbe Länge haben!"

# ============ Ausgabe ==================

print(f"V = {V},  W = {W}")
for i in range(len(p)):
    print(f"p_{i+1}={ps.p_list[i]}")
print("-"*50)
print("Ein Würfel:")
# exakt - ein-Würfel
# exakte Funktion bauen
gew = ps.baue_P_gewinn(tuple(p))
PA, PB, PU = gew(V, W)

print(f"exakt:  P_A = {float(PA):.4f},  P_B = {float(PB):.4f},  P_U = {float(PU):.4f}\n")
print("-"*50)
# Ein-Würfel-Simulation mit Wald-CI
cis1 = ps.simuliere_duell_1_wuerfel(V, W, p, N=sim, alpha=0.05, seed=42)
# Punkt-Schätzung und CI entpacken
pA_hat1, (pA_low1, pA_high1) = cis1['P_A']
pB_hat1, (pB_low1, pB_high1) = cis1['P_B']
pU_hat1, (pU_low1, pU_high1) = cis1['P_U']

print("numerisch - Ein-Würfel und Zwei-Würfel (gerundet auf 4 Nachkommastellen):")
print(f"jeweils {sim} Simulationen\n")
print("Ein Würfel:")
print(f"P_A = {pA_hat1:.4f}, 95%-CI = [{pA_low1:.4f}, {pA_high1:.4f}]")
print(f"P_B = {pB_hat1:.4f}, 95%-CI = [{pB_low1:.4f}, {pB_high1:.4f}]")
print(f"P_U = {pU_hat1:.4f}, 95%-CI = [{pU_low1:.4f}, {pU_high1:.4f}]\n")

# Zwei-Würfel-Simulation mit Wald-CI
cis2 = ps.simuliere_duell_2_wuerfel(V, W, p, p, N=sim, alpha=0.05)
pA_hat2, (pA_low2, pA_high2) = cis2['P_A']
pB_hat2, (pB_low2, pB_high2) = cis2['P_B']
pU_hat2, (pU_low2, pU_high2) = cis2['P_U']

print("Zwei Würfel:")
print(f"P_A = {pA_hat2:.4f}, 95%-CI = [{pA_low2:.4f}, {pA_high2:.4f}]")
print(f"P_B = {pB_hat2:.4f}, 95%-CI = [{pB_low2:.4f}, {pB_high2:.4f}]")
print(f"P_U = {pU_hat2:.4f}, 95%-CI = [{pU_low2:.4f}, {pU_high2:.4f}]")
```
